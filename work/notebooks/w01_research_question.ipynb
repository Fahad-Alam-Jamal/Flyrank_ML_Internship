## 3. Quick look at the data (2-3 real numbers)

The starter dataset has enough pages and measurable decline signal to make refresh scoring worth the next 7 weeks. Below I compute the sample size, the rate of declining pages, and how many visible pages are still moving down.


```python
import pandas as pd
from pathlib import Path
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent.parent
df = pd.read_csv(root / 'data' / 'raw' / 'content_refresh_anonymized.csv')

n_rows = len(df)
declining_rate = (df['trend_direction'] == 'down').mean()
visible = df['impressions_90d'] >= 500
visible_count = visible.sum()
visible_declining = ((df['trend_direction'] == 'down') & visible).sum()
visible_top20_declining = ((df['trend_direction'] == 'down') & visible & (df['avg_position'] <= 20)).sum()
old_visible = ((df['content_age_days'] >= 180) & visible).sum()

print('Starter rows:', n_rows)
print('Declining rate:', f'{declining_rate:.3f}')
print('Visible pages (impressions_90d >= 500):', visible_count)
print('Visible pages that are declining:', visible_declining)
print('Visible pages with position <= 20 that are declining:', visible_top20_declining)
print('Visible pages older than 180 days:', old_visible)
```
